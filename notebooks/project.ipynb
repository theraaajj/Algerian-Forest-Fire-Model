#### 1. Data Cleaning


```python
# checking for missing values in columns
data.isnull().sum()
```

```python
# checking for missing values in rows
data[data.isnull().any(axis=1)]
```

> This shows the row which has the null value

SO! SO! SO!
The dataset is converted into two sets based on region from 122th index(row). 

To resolve this issue we can make a new column based on the Region

1: Bejaia Region dataset

2: Sidi-Bel Abbes Region Dataset

We add a new column "Region"

```python
data.loc[:122, "Region"]=0
data.loc[122:, "Region"]=1
data.head()
```

```python
data.info()
```

> we need not store the region as float, it'll take more space
> so we convert them to int

```python
data[["Region"]] = data[["Region"]].astype(int)
data.info()
```

> NICE!

```python
data.isnull().sum()
```

```python
# let's just drop the row 122, and 123 (unnecessary)
data = data.dropna().reset_index(drop=True)
```

```python
data.isnull().sum()
```

> null values eliminated

> But are the indices maintained?

```python
data.iloc[[122]]
```

> YESS!

> But what we see here is that the column names are repeated! Let's just eliminate this row

```python
data = data.drop(122).reset_index(drop=True)
```

> We're almost done!

> Let's see a few not so imp things

```python
data.columns
```

> some of them have spaces

```python
# removing spaces in the column names
data.columns = data.columns.str.strip()
data.columns
```

```python
data.info()
```

> we don't need these Dtype as object. We should change it to int or float

##### Changing the columns into int type

```python
data[['month', 'day', 'year', 'Temperature', 'RH', 'Ws']] = data[['month', 'day', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
data.info()
```

```python
data[['Classes']] = data[['Classes']].astype(str)
data.info()
```

##### Changing the other columns to float

```python
columnsToBeConverted = [features for features in data.columns if data[features].dtypes=='O']
```

```python
for x in columnsToBeConverted:
    if x!='Classes':
        data[x] = data[x].astype(float)
```

```python
data.info()
```

```python
data.describe()
```

> **This was all for Data Cleaning here!**

Let's save the cleaned dataset

```python
data.to_csv('Algerian_forest_fires_cleaned_dataset.csv', index=False)
```

> as we can see, some of them are skewed

Skewness measures the asymmetry of a data distribution. If you visualize your data using a histogram or frequency curve, skewness indicates which side of the distribution is more stretched out or elongated than the other, and which side has a tail.

```python
# checking percentage of fire and not fire categories
# data_copy['Classes'].value_counts(normalize=True)*100
percentage = data_copy['Classes'].value_counts(normalize=True)*100
```

```python
# plotting piechart
classlabels = ["Fire", "Not Fire"]
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels, autopct='%1.1f%%')
plt.title("PieChart of Classes")
plt.show()
```
